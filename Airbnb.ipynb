```python
def remove_outliers(df, group_col, target_col, quantile=0.99):
    thresholds = df.groupby(group_col)[target_col].quantile(quantile).to_dict()
    return df[df.apply(lambda x: x[target_col] <= thresholds.get(x[group_col], x[target_col]), axis=1)]

df_clean = remove_outliers(df, 'room_type', 'price')
print(f"After removing outliers: {len(df_clean)} listings remaining.")
```
